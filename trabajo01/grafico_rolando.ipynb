```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1️⃣ Cargar los datos desde Excel
df = pd.read_excel("./data_Leads enero-julio 2025.xlsx")

# 2️⃣ Agrupar los datos: contar cuántos leads hay por tipo y nivel de interés
tabla = df.groupby(["Tipo Persona", "Nivel De Interes"]).size().unstack(fill_value=0)

# 3️⃣ Crear el gráfico de barras horizontales
tabla.plot(kind="barh", figsize=(10,6), colormap="viridis")

# 4️⃣ Personalizar el gráfico
plt.title("Tipo de Lead vs Nivel de Interés", fontsize=14)
plt.xlabel("Cantidad de Leads", fontsize=12)
plt.ylabel("Tipo de Lead", fontsize=12)
plt.legend(title="Nivel de Interés")
plt.tight_layout()

# 5️⃣ Mostrar
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt


# 1️⃣ Cargar los datos desde Excel
df = pd.read_excel("./data_Leads enero-julio 2025-1.xlsx")

# 2️⃣ Filtrar los niveles de interés
niveles_seleccionados = ['compró','alto', 'medio', 'bajo']

# 3️⃣ Filtrar el DataFrame
df_filtrado = df[df['Nivel De Interes'].isin(niveles_seleccionados)]


# 4️⃣ Crear la tabla de conteo
tabla_mascotas = df_filtrado.groupby(["Tiene mascotas (si,no)", "Nivel De Interes"]).size().unstack(fill_value=0)


# 5️⃣ Crear el gráfico
tabla_mascotas.plot(kind='bar', stacked=True, figsize=(4, 6))

plt.title("Mascotas vs Nivel de interés")
plt.xlabel("Tiene mascotas (si/no)")
plt.ylabel("Número de Leads")
plt.xticks(rotation=0)
plt.legend(title="Nivel de interés")
plt.tight_layout() 
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo Excel
file_path = "data_Leads enero-julio 2025-1.xlsx"
df = pd.read_excel(file_path)

# Verificamos las columnas disponibles (puedes descomentar esto para inspeccionar)
# print(df.columns)

# Agrupar por el área y el nivel de interés (ajusta el nombre de las columnas si difieren)
df_grouped = df.groupby(
    # ['en_que_area_del_departamento_cree_usted_que_optimizamos_el_metro_cuadrado', 'Nivel De Interes']
    ['en_que_area_del_departamento_cree_usted_que_optimizamos_el_metro_cuadrado', 'Proyectos Relacionados']
).size().unstack(fill_value=0)

# Crear el gráfico apilado (stacked bar)
ax = df_grouped.plot(
    kind='barh',
    stacked=True,
    figsize=(10, 6),
    colormap='viridis'
)

# Títulos y etiquetas
plt.title(
    'OPTIMIZACIÓN DE METRO CUADRADO VS NIVEL DE INTERÉS EN BASE A LOS PROYECTOS RELACIONADOS',
    fontsize=12,
    fontweight='bold',
    loc='center',
    pad=20
)
plt.xlabel('Número de Proyectos Relacionados', fontsize=10)
plt.ylabel('Áreas del Departamento', fontsize=10)

# Personalizar leyenda
plt.legend(title='Nivel de Interés', bbox_to_anchor=(1.05, 1), loc='upper left')

# Mejorar espaciado
plt.tight_layout()

# Mostrar gráfico
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

# Leer el archivo Excel
file_path = "data_Leads enero-julio 2025-1.xlsx"
df = pd.read_excel(file_path)

df_grouped = df.groupby(
    ['en_que_area_del_departamento_cree_usted_que_optimizamos_el_metro_cuadrado', 'Proyectos Relacionados']
).size().unstack(fill_value=0)

df_grouped = df_grouped[df_grouped.sum(axis=1) > 5]

ax = df_grouped.plot(
    kind='barh',
    stacked=True,
    figsize=(10, 6),
    colormap='viridis'
)

plt.title(
    'AREA DE OPTIMIZACIÓN DEL DEPARTAMENTO VS EN BASE A LOS PROYECTOS RELACIONADOS',
    fontsize=12,
    fontweight='bold',
    loc='center',
    pad=20
)
plt.xlabel('Número de Proyectos Relacionados', fontsize=10)
plt.ylabel('Áreas del Departamento', fontsize=10)

plt.legend(title='Proyectos Relacionados', bbox_to_anchor=(1.05, 1), loc='upper left')

plt.tight_layout()

plt.show()
```
