```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Cargar el archivo csv
dataF = pd.read_csv("./data_arroz.csv" , encoding="latin-1")
# dataF = pd.read_csv("./data_arroz.csv", sep=';', encoding='utf-8')

# Ver las primeras filas para confirmar que se cargó bien
print(dataF.head(5))
# print(dataF.columns)
```

```python

# Convertir el DataFrame a un array
array_arroz = dataF.values
print(f"El array tiene: {array_arroz}")
```

```python
# dimensiones del array 2
print("el array tiene  dimensiones: ", array_arroz.ndim)

# el tamaño de cada dimension primera dimension filas 27 , segunda dimension columnas 4
print("el tamaño de cada dimension: ", array_arroz.shape)

# la cantidad de registros del arreglo
print("la cantidad de registros del arreglo es: ", array_arroz.size)
```

```python


# seleccionar solo las columnas precio y demanda
array_arroz_2 = array_arroz[:,2:]

print(f"El array tiene: {array_arroz_2}")

# dimensiones del array 2
print("el array tiene  dimensiones: ", array_arroz_2.ndim)

# el tamaño de cada dimension primera dimension filas 27 , segunda dimension columnas 2
print("el tamaño de cada dimension: ", array_arroz_2.shape)

# la cantidad de registros del arreglo
print("la cantidad de registros del arreglo es: ", array_arroz_2.size)

```

```python
# seleccionar solo las columnas precio y demanda
# maxtriz x =  precio 
# maxtriz y =  demanda

x = array_arroz_2[:,0]
y = array_arroz_2[:,1]


print(f"El array tiene: {x}")

# dimensiones del array
print("el array tiene  dimensiones: ", x.ndim)

# el tamaño de cada dimension 
print("el tamaño de cada dimension: ", x.shape)

# la cantidad de registros del arreglo
print("la cantidad de registros del arreglo es: ", x.size)

```

```python
# estadisticas de precio
print("estadisticas de precio")
print(f"precio minimo: {x.min()}")
print(f"precio maximo: {x.max()}")
print(f"precio promedio: {x.mean()}")

print("\n")

# estadisticas de demanda
print("estadisticas de demanda")
print(f"demanda minima: {y.min()}")
print(f"demanda maxima: {y.max()}")
print(f"demanda promedio: {y.mean()}")
```

```python
# CALCULO DE LOS COEFICIENTES DE REGRESIO LINEAL
# Y(target/objetivo) es la demanda semanla de unidades vendidas
# X es el precio

x_ = np.column_stack((np.ones(len(x)), x))

print(f"el arreglo de la matriz x es = {x_}")
```
