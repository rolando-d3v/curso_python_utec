## <font color='lightcoral'>**Ejercicio 1:** Método NO gráfico Univariado</font> (2 Puntos)

Analiza un dato categórico de tu dataset utilizando el método no gráfico.

```python
import pandas as pd

df = pd.read_csv('https://raw.githubusercontent.com/germaingarcia/diplomadoFiles/main/Spotify_dataset.csv')

# Ejercicio 1: Método no gráfico univariado (dato categórico)
conteo_generos = df['genre'].value_counts()
porcentaje_generos = df['genre'].value_counts(normalize=True) * 100

print("Frecuencia de cada género:")
print(conteo_generos)

print("\nPorcentaje de cada género (%):")
print(porcentaje_generos.round(2))
```
